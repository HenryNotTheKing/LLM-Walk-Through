## 3. 文档编写规范

本项目为了兼顾**教学直观性**与**源码生产力**，采用了以下规范：

- **Notebook 化**：所有技术文档均以 `.ipynb` 形式呈现，包含“原理讲解 -> 交互代码 -> 源码映射”。
- **Code Map**：每个 Notebook 结尾都必须将概念映射到本仓库的具体文件、测试和配置中。
- **References**：严格按 Paper, Blog, Code 三类列出参考资料。
