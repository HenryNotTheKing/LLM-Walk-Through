# 整体代码模块导航书与阅读说明

这份 `modules` 下的各个 notebook 是配合本 LLM Walk-Through 的互动式课堂。在这里，你将逐步地学习大模型所需的各式零部件是如何抛弃各种臃长功能，变成最高效形式的。

下面是这个目录下各个部分的教学路径索引：

**1. 基础架构部件**
- 归一化层 `norm/01_rmsnorm.ipynb`: 去除偏置项和中心化限制，大幅节省显存的快速归一化技术
- 位置编码 `position/01_rope.ipynb`: 为什么基于旋转复数的位置编码可以赋予网络绝佳推导相对位置和泛化的能力
- 激活函数网络 `ffn/01_swiglu.ipynb`: Gated Linear Unit，替代传统 GELU 进行非线性提纯的双并联计算流

**2. 注意力引擎与模型演化**
- 注意力模块与演化 `attention/03_attention_mha.ipynb` 和 `attention/04_gqa_qk_norm_rope.ipynb`: 从多头注意力（MHA）讲到分组共享查询的 GQA，配合 QK-Norm 缓解深层激活爆炸
- GPT2基石模型 `model/02_gpt2_baseline.ipynb` 与 `model/04_code_causal_lm.ipynb`: 如何用以上的部件组合、替换掉古董的 GPT2 核心并套上最新代码实现

**3. 数据流与预存续工程（重点内容）**
- 分词器训练 `tokenizer/`相关内容：如何从生文本抽出高质量的词表字典。
- 存档保存与加载系统 `utils/01_checkpoint.ipynb`：大模型训练常常出车祸（停电/OOM等），因此需要完善并且稳妥的容灾存储与进度恢复系统，这里包含了我们的具体实现逻辑。

阅读顺序可以随意，建议优先跟着自己困惑的核心结构入手去跑一跑。在里面的每一个 notebook 里都有相关的测试案例、详细讲解和在 `core/` 下具体对应代码的使用。

## 结构约束

`docs/modules/` 的一级目录应基本对应 `core/` 的一级包。下面的代码块会扫描两边目录并打印映射状态，后续新增模块时可以直接运行检查。

```python
from __future__ import annotations

from pathlib import Path

ROOT = Path.cwd()
while ROOT.name and not (ROOT / 'pyproject.toml').exists():
    ROOT = ROOT.parent

core_dirs = sorted(
    p.name for p in (ROOT / 'core').iterdir()
    if p.is_dir() and not p.name.startswith('__')
)
doc_dirs = sorted(
    p.name for p in (ROOT / 'docs' / 'modules').iterdir()
    if p.is_dir() and not p.name.startswith('__')
)

print('core modules:', core_dirs)
print('doc modules :', doc_dirs)
print('missing docs:', sorted(set(core_dirs) - set(doc_dirs)))
print('extra docs  :', sorted(set(doc_dirs) - set(core_dirs)))
```
