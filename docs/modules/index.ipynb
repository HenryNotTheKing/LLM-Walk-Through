## 结构约束

`docs/modules/` 的一级目录应基本对应 `core/` 的一级包。下面的代码块会扫描两边目录并打印映射状态，后续新增模块时可以直接运行检查。

```python
from __future__ import annotations

from pathlib import Path

ROOT = Path.cwd()
while ROOT.name and not (ROOT / 'pyproject.toml').exists():
    ROOT = ROOT.parent

core_dirs = sorted(
    p.name for p in (ROOT / 'core').iterdir()
    if p.is_dir() and not p.name.startswith('__')
)
doc_dirs = sorted(
    p.name for p in (ROOT / 'docs' / 'modules').iterdir()
    if p.is_dir() and not p.name.startswith('__')
)

print('core modules:', core_dirs)
print('doc modules :', doc_dirs)
print('missing docs:', sorted(set(core_dirs) - set(doc_dirs)))
print('extra docs  :', sorted(set(doc_dirs) - set(core_dirs)))
```
