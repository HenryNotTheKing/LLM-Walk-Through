# Causal Multi-Head Self-Attention (因果多头自注意力)

> 源码对应：[`core/attention/mha.py`](../../core/attention/mha.py)

注意力机制是 Transformer 的灵魂。在 GPT 这样的生成式模型中，我们使用的是**因果 (Causal)** 注意力，即每个位置只能“看到”它之前的 Token。

## 1. 数学原理

对于投影得到的 Query ($Q$), Key ($K$), Value ($V$)：

$$\mathrm{Attn}(Q,K,V) = \mathrm{softmax}\!\left(\frac{QK^\top}{\sqrt{d_h}} + M\right) V$$

其中 $M$ 是 **因果掩码 (Causal Mask)**，它将右上三角（未来位置）的值设为 $-\infty$，从而在 softmax 后权重变为 0。

---

## 2. 三种实现路径演示

本项目提供了两种（未来会增加更多）实现方式：`sdpa` 和 `eager`。

### 2.1 Eager 实现 (教学版)
这种方式手动处理 Mask 和 Softmax，最适合理解原理。

源码对应：`CausalSelfAttention` 中的 [`eager` 分支](../../core/attention/mha.py#L85)


```python
import torch
import torch.nn.functional as F

B, T, C = 1, 4, 32 # Batch, Time, Channels
q = torch.randn(B, T, C)
k = torch.randn(B, T, C)

# 1. 计算点积分数
att = (q @ k.transpose(-2, -1)) * (1.0 / (C**0.5))

# 2. 创建并应用因果掩码
mask = torch.tril(torch.ones(T, T)).view(1, 1, T, T)
att = att.masked_fill(mask[:, :, :T, :T] == 0, float('-inf'))

# 3. Softmax
att = F.softmax(att, dim=-1)

print("因果注意力权重矩阵 (Softmax 后):")
print(att[0, 0] if att.ndim == 4 else att[0]) # 观察其下三角特性
```

### 2.2 SDPA 实现 (工程加速版)
PyTorch 2.0+ 提供的 `scaled_dot_product_attention` 会自动根据硬件选择最快的内核（如 FlashAttention）。

源码对应：`CausalSelfAttention` 中的 [`sdpa` 分支](../../core/attention/mha.py#L74)


```python
# 在 PyTorch 中一行搞定，且显存更省、速度更快
# is_causal=True 会自动处理上述所有掩码逻辑
q_m = q.view(B, 1, T, C) # 模拟多头维度 (B, H, T, d_h)
k_m = k.view(B, 1, T, C)
v_m = torch.randn(B, 1, T, C)

y = F.scaled_dot_product_attention(q_m, k_m, v_m, is_causal=True)
print(f"SDPA 输出形状: {y.shape}")
```

## 3. 几个关键工程细节

- **`c_attn` 优化**：在源码中，我们使用一个大 `Linear` 层同时投射出 Q, K, V，然后通过 `split` 拆分，这比分三次投影更快。
- **初始化缩放**：GPT-2 论文指出，在残差路径上的输出投影层（`c_proj`）权重应该按 $1/\sqrt{2N}$ 进行缩放（$N$ 是层数），这有助于深层网络的训练。
