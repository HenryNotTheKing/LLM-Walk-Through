## V1 · 现代主干替换 (进行中)

我们将 GPT-2 中的旧组件逐个替换为现代主流 LLM (如 LLaMA, Mistral) 使用的方案：

| 模块 | 替换为 | 核心收益 |
| --- | --- | --- |
| **位置编码** | **RoPE** (旋转位置编码) | 更好的长度外推能力，消除固定窗口限制 |
| **归一化层** | **RMSNorm** | 训练更稳定（尤其是深层网络），计算更高效 |
| **激活函数** | **SwiGLU** | 目前公认的 SOTA 激活函数，提升模型容量 |
| **注意力优化** | **GQA** (分组查询注意力) | 大幅降低推理时的 KV Cache 显存占用 |

---

## V2 · 对齐与强化学习 (愿景)

- **SFT (Supervised Fine-Tuning)**：在指令数据集上进行全步微调。
- **DPO / SimPO**：无需奖励模型的直接偏好对齐。
- **RLHF (GRPO)**：借鉴 DeepSeek-R1 的强化学习方案，提升模型的推理能力。
